```python
import nltk
from nltk.corpus import reuters
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
import numpy as np

# Download required NLTK data
nltk.download('reuters')
nltk.download('punkt')

documents = reuters.fileids()
corpus = [reuters.raw(doc_id) for doc_id in documents]

print("Number of documents:", len(documents))
print("Number of text documents:", len(corpus))
print("Total characters:", sum(len(doc) for doc in corpus))

print("\nFirst document ID:", documents[0])
print("\nSample document:")
print(corpus[0][:1000])
```

### Task1: How many documents are there? What does a sample document look like?

Number of documents: 10788
It is Reuters news article, include information about business, economics, financial markets, or other news topics. 

```python
bow_vectorizer = CountVectorizer(
    lowercase=True,
    token_pattern=r'\b\w+\b',
    min_df=1
)

# Generate BoW representation
X_bow = bow_vectorizer.fit_transform(corpus)

# Create a DataFrame for better visualization
bow_df = pd.DataFrame(
    X_bow[:5].toarray(),
    columns=bow_vectorizer.get_feature_names_out()
)

print("\nBag-of-Words Representation:")
print(bow_df)

print("\nNumber of documents:", X_bow.shape[0])
print("Vocabulary size:", len(bow_vectorizer.vocabulary_))
print("BoW matrix shape:", X_bow.shape)

print("\nFirst 20 vocabulary terms:")
print(bow_vectorizer.get_feature_names_out()[:20])

sorted_vocabulary = sorted(
    bow_vectorizer.vocabulary_.items(),
    key=lambda item: item[1]
)

print("\nFirst 20 terms sorted by index:")
print(sorted_vocabulary[:20])
```

### Task 2: How might you sort this to see terms in order?

I use CountVectorizer to count the words across a large number of articles. I then configure the tool to make the output easier to work with—for instance, by converting all text to lowercase (e.g., changing "Apple" to "apple") so that the same word isn't counted multiple times. In the third step, I have the tool process all the articles, identify every unique word, and organize them into a large table. The tool then reveals the total number of distinct words, which is referred to as the "vocabulary."

```python
# Create and configure TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer(
    lowercase=True,
    token_pattern=r'\b\w+\b',
    min_df=1
)

# Generate TF-IDF representation
X_tfidf = tfidf_vectorizer.fit_transform(corpus)

# Create a DataFrame for better visualization
tfidf_df = pd.DataFrame(
    X_tfidf[:5].toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("\nTF-IDF Representation:")
print(tfidf_df)

print("\nTF-IDF vocabulary size:",
      len(tfidf_vectorizer.vocabulary_))

print("TF-IDF matrix shape:", X_tfidf.shape)

print("\nBoW matrix shape:", X_bow.shape)
print("TF-IDF matrix shape:", X_tfidf.shape)

doc_index = 0

bow_indices = X_bow[doc_index].nonzero()[1]
tfidf_indices = X_tfidf[doc_index].nonzero()[1]

print("\nNumber of non-zero BoW features:",
      len(bow_indices))

print("Number of non-zero TF-IDF features:",
      len(tfidf_indices))

print("\nFirst 15 TF-IDF vocabulary terms:")
print(tfidf_vectorizer.get_feature_names_out()[:15])

print("\nFirst 15 non-zero TF-IDF features in document 1:")

for idx in tfidf_indices[:15]:
    term = tfidf_vectorizer.get_feature_names_out()[idx]
    score = X_tfidf[doc_index, idx]
    print(f"{term}: {score:.4f}")
```

```python
tfidf_stop_vectorizer = TfidfVectorizer(
    lowercase=True,
    min_df=1,
    stop_words='english'
)

tfidf_terms = tfidf_vectorizer.get_feature_names_out()

X_tfidf_stop = tfidf_stop_vectorizer.fit_transform(corpus)

stop_terms = tfidf_stop_vectorizer.get_feature_names_out()

print("Original TF-IDF vocabulary size:", len(tfidf_terms))
print("TF-IDF vocabulary after stop word removal:", len(stop_terms))

print("\nOriginal TF-IDF matrix shape:", X_tfidf.shape)
print("TF-IDF matrix after stop word removal:", X_tfidf_stop.shape)

original_features = X_tfidf[0].nnz
filtered_features = X_tfidf_stop[0].nnz

print("\nFirst document non-zero features before removal:", original_features)
print("First document non-zero features after removal:", filtered_features)
print("Number of features removed:", original_features - filtered_features)

original_vocab_size = len(tfidf_terms)
new_vocab_size = len(stop_terms)

reduction = (
    (original_vocab_size - new_vocab_size) / original_vocab_size * 100
)

print("\nVocabulary reduction (%):", round(reduction, 2))

stop_indices = X_tfidf_stop[0].nonzero()[1]

print("\nFirst 15 remaining features:")
for idx in stop_indices[:15]:
    print(stop_terms[idx], ":", round(X_tfidf_stop[0, idx], 4))
```

### Task 4: How much does the vocabulary size change? How does the number of features per document change? What’s the impact on the representation quality?

Removing stop words reduces the number of distinct terms in the vocabulary. After removing stop words, the vocabulary decreased to 30,627 terms.

For the first document, the number of non-zero features decreased from 360 to 269.

By removing frequently occurring words, the number of terms to process is reduced, allowing for a focus on more meaningful words within the articles. These words help identify the main topic of each article.

```python
def process_new_document(document, vectorizer):
    """
    Process a new document using a trained vectorizer.

    Args:
        document (str): Text to process
        vectorizer: Trained CountVectorizer or TfidfVectorizer

    Returns:
        sparse matrix: Document vector
    """
    # Transform the new document using the existing vocabulary
    document_vector = vectorizer.transform([document])

    return document_vector


new_document = """
Financial markets rose today as investors reacted to
economic growth and company earnings. Stock prices increased,
while oil prices remained stable. Analysts expect the economy
to continue growing if inflation stays under control.
"""

new_bow = process_new_document(new_document, bow_vectorizer)

new_tfidf = process_new_document(new_document, tfidf_vectorizer)

new_tfidf_stop = process_new_document(
    new_document, tfidf_stop_vectorizer
)

print("New document BoW shape:", new_bow.shape)
print("New document TF-IDF shape:", new_tfidf.shape)
print("New document TF-IDF without stop words shape:",
      new_tfidf_stop.shape)

print("\nBoW non-zero features:", new_bow.nnz)
print("TF-IDF non-zero features:", new_tfidf.nnz)
print("TF-IDF non-zero features after stop word removal:",
      new_tfidf_stop.nnz)

bow_new_indices = new_bow.nonzero()[1]
tfidf_new_indices = new_tfidf.nonzero()[1]
stop_new_indices = new_tfidf_stop.nonzero()[1]

print("\nTerms found by BoW:")
print([bow_vectorizer.get_feature_names_out()[i]
       for i in bow_new_indices])

print("\nTerms found by TF-IDF:")
print([tfidf_vectorizer.get_feature_names_out()[i]
       for i in tfidf_new_indices])

print("\nTerms found after stop word removal:")
print([tfidf_stop_vectorizer.get_feature_names_out()[i]
       for i in stop_new_indices])
```
